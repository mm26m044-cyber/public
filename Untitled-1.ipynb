```python
"""
fetch_data.py

Step 1 of Band Explorer: fetch a material's electronic structure data
from the Materials Project and save it locally as JSON.

This is the ONLY part of the project that needs the internet or an
API key. Every later step (Varshni temperature correction, strain,
effective-mass extraction, plotting) will work offline on the JSON
files this script produces.

Usage:
    python fetch_data.py
"""

import os
import json

from dotenv import load_dotenv
from monty.json import MontyEncoder
from mp_api.client import MPRester

# Silicon's Materials Project ID — our first test material.
MATERIAL_ID = "mp-149"
OUTPUT_DIR = os.path.join("data", "raw")


def get_api_key() -> str:
    """Load the Materials Project API key from a local config.env file."""
    load_dotenv(dotenv_path="config.env")  # reads config.env in the project root
    api_key = os.environ.get("MP_API_KEY")

    if not api_key:
        raise RuntimeError(
            "MP_API_KEY not found.\n"
            "Copy config.env.example to config.env and paste in your key, "
            "or set it as an environment variable directly.\n"
            "Get a free key at: https://next-gen.materialsproject.org/api"
        )
    return api_key


def fetch_material(material_id: str) -> dict:
    """Fetch structure, band structure, and DOS for one material ID."""
    with MPRester(get_api_key()) as mpr:
        structure = mpr.get_structure_by_material_id(material_id)

        try:
            bandstructure = mpr.get_bandstructure_by_material_id(material_id)
            dos = mpr.get_dos_by_material_id(material_id)
        except Exception as e:
            if "equivalent_labels" in str(e):
                raise RuntimeError(
                    "Known mp-api bug: band structure fetch failed with a "
                    "Pydantic validation error. Fix: make sure your "
                    "installed versions exactly match requirements.txt "
                    "(mp-api==0.46.4, emmet-core==0.87.1) — "
                    "run: pip install -r requirements.txt --force-reinstall"
                ) from e
            raise

    return {
        "material_id": material_id,
        "structure": structure,
        "bandstructure": bandstructure,
        "dos": dos,
    }


def save_raw(data: dict, output_dir: str = OUTPUT_DIR) -> str:
    """
    Save the fetched pymatgen objects as JSON.

    MontyEncoder is pymatgen's own JSON encoder — it knows how to
    serialize Structure/BandStructure/Dos objects, which plain
    json.dump() cannot do on its own.
    """
    os.makedirs(output_dir, exist_ok=True)
    out_path = os.path.join(output_dir, f"{data['material_id']}.json")

    with open(out_path, "w") as f:
        json.dump(data, f, cls=MontyEncoder)

    return out_path


def main():
    print(f"Fetching {MATERIAL_ID} (Silicon) from Materials Project...")
    data = fetch_material(MATERIAL_ID)

    gap = data["bandstructure"].get_band_gap()
    print(f"  Band gap: {gap['energy']:.3f} eV "
          f"({'direct' if gap['direct'] else 'indirect'})")

    out_path = save_raw(data)
    print(f"Saved raw data to: {out_path}")


if __name__ == "__main__":
    main()
```

```python
import json

with open("data/raw/mp-149.json") as f:
    raw = json.load(f)

print(type(raw))
print(list(raw.keys())[:20])
print(raw.get("@class"), raw.get("@module"))

from monty.json import MontyDecoder

bs = MontyDecoder().process_decoded(raw["bandstructure"])
print(type(bs))

print(raw["bandstructure"].keys())
print(raw["bandstructure"].get("@class"))

gap_info = bs.get_band_gap()
cbm = bs.get_cbm()
vbm = bs.get_vbm()

print(f"Metal: {bs.is_metal()}")
print(f"Gap: {gap_info['energy']:.3f} eV ({'direct' if gap_info['direct'] else 'indirect'})")
print(f"Transition: {gap_info['transition']}")
print(f"VBM: {vbm['energy']:.3f} eV at k = {vbm['kpoint'].frac_coords}, label = {vbm['kpoint'].label}")
print(f"CBM: {cbm['energy']:.3f} eV at k = {cbm['kpoint'].frac_coords}, label = {cbm['kpoint'].label}")
```
